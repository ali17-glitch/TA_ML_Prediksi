### 1. Perbaikan Model
Dalam pengembangan ini, dilakukan perbaikan preprocessing dan penambahan fitur guna meningkatkan kualitas prediksi:
- **a) Perbaikan Preprocessing:** Dilakukan pembersihan data, penanganan outlier, imputasi nilai kosong (jika ada), serta standardisasi skala data menggunakan `StandardScaler` agar algoritma jarak (seperti Linear Regression/Gradient Boosting) tidak terpengaruh oleh perbedaan satuan variabel.

**Mengapa dilakukan?** Standardisasi dan rekayasa fitur dilakukan untuk mengurangi bias model, meningkatkan akurasi generalisasi, serta memastikan bahwa model tidak mengalami overfitting saat memproses data kabupaten/kota baru.

### 2. Training Model
Dilatih menggunakan minimal **2 algoritma Machine Learning** (Regresi):
1. **Random Forest Regressor** (Model Ensemble berbasis pohon keputusan yang kokoh)
2. **Linear Regression** (Model parametrik dasar sebagai baseline untuk perbandingan)
3. **Gradient Boosting Regressor** (Model boosting adaptif untuk performa optimal)

### 3. Evaluasi & Perbandingan Model
Karena kasus ini merupakan kasus Regresi, metrik evaluasi yang digunakan adalah:
- **MAE (Mean Absolute Error)**: Mengukur rata-rata selisih absolut antara nilai aktual dan prediksi.
- **RMSE (Root Mean Squared Error)**: Memberikan penalti lebih besar untuk error yang nilainya besar.
- **R² (Coefficient of Determination)**: Mengukur seberapa baik variasi tingkat kemiskinan dapat dijelaskan oleh fitur-fitur model.

```python
# Training Model
# Modeling
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor

# Evaluasi
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

import numpy as np
```

```python
# Membuat Fungsi Evaluasi
def evaluasi_model(y_true, y_pred):

    mae = mean_absolute_error(y_true, y_pred)
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_true, y_pred)

    print("MAE :", mae)
    print("MSE :", mse)
    print("RMSE:", rmse)
    print("R2  :", r2)
```

Penjelasan metrik:

- MAE
Mengukur rata-rata selisih absolut antara prediksi dan nilai sebenarnya.

- MSE
Mengukur rata-rata kuadrat kesalahan prediksi.

- RMSE
Akar dari MSE sehingga kembali ke satuan asli data.

- R²
Menunjukkan seberapa baik model menjelaskan variasi data.

Nilai R² mendekati 1 berarti model sangat baik.

```python
# Training Model Linear Regression
model_lr = LinearRegression()

model_lr.fit(X_train, y_train)

y_pred_lr = model_lr.predict(X_test)
```

Penjelasan

Linear Regression memodelkan hubungan linear antara variabel input dan target.

Model ini digunakan sebagai baseline model untuk melihat hubungan dasar antar variabel.

```python
#Evaluasi Linear Regression
print("Hasil Evaluasi Linear Regression")

evaluasi_model(y_test, y_pred_lr)
```

Interpretasi

Model mampu menjelaskan sekitar 67% variasi tingkat kemiskinan.

Artinya indikator sosial ekonomi cukup menjelaskan tingkat kemiskinan.

```python
# Training Model random Forest Regressor
model_rf = RandomForestRegressor(
    n_estimators=200,
    random_state=42
)

model_rf.fit(X_train, y_train)

y_pred_rf = model_rf.predict(X_test)
```

Penjelasan

Random Forest merupakan algoritma ensemble learning yang menggunakan banyak decision tree.

Kelebihan:

mampu menangkap hubungan non-linear

lebih tahan terhadap noise data

memiliki performa tinggi pada banyak kasus regresi

```python
# Evaluasi Random Forest
print("Hasil Evaluasi Random Forest")

evaluasi_model(y_test, y_pred_rf)
```

### **1. Bagaimana model membantu Smart City?**
Model ini berkontribusi langsung pada pilar **Smart Governance** (Tata Kelola Cerdas) dan **Smart Society** (Masyarakat Cerdas) dengan menyajikan analisis prediktif berbasis bukti (*data-driven policy*). Dengan model ini, pengambilan keputusan tidak lagi bergantung pada dugaan intuitif, melainkan pada variabel riil penentu kemiskinan seperti tingkat pendidikan (RLS/HLS) dan daya beli (Pengeluaran Per Kapita).

### **2. Apa manfaat sistem bagi pemerintah/masyarakat?**
- **Bagi Pemerintah:** Membantu Bappeda dan Dinas Sosial memetakan potensi kerawanan kemiskinan antar wilayah secara real-time, sehingga perencanaan program jaminan sosial dan pembangunan ekonomi daerah menjadi lebih terukur dan tepat sasaran (*targeted programming*).
- **Bagi Masyarakat:** Mempercepat penyaluran bantuan sosial secara adil, transparan, dan meminimalkan ketimpangan infrastruktur dasar (pendidikan/kesehatan) bagi daerah-daerah yang paling membutuhkan.

### **3. Apa risiko jika prediksi salah?**
- **Exclusion Error (Kekeliruan Pengecualian):** Warga miskin yang sangat membutuhkan bantuan terprediksi sebagai "kondisi aman", sehingga kehilangan hak atas bantuan sosial yang berdampak pada peningkatan kemiskinan ekstrem.
- **Inclusion Error (Kekeliruan Pemasukan):** Warga mampu terprediksi sebagai "miskin", sehingga terjadi pemborosan anggaran daerah dan ketidakefisienan alokasi logistik pemerintah.
- **Over-reliance:** Risiko ketergantungan berlebih terhadap teknologi digital tanpa dilakukannya validasi faktual atau cek lapangan secara berkala.

### **4. Bagaimana menjaga privasi data?**
- **Anonimisasi dan Agregasi:** Menghapus seluruh data identitas langsung warga (seperti NIK, nama, dan alamat rumah) dan hanya menampilkan data agregat di tingkat kabupaten/kota.
- **Kepatuhan Keras UU PDP No. 27 Tahun 2022:** Menerapkan pembatasan akses data menggunakan *Role-Based Access Control* (RBAC) dan enkripsi pangkalan data untuk mencegah penyalahgunaan informasi sensitif oleh pihak luar.
